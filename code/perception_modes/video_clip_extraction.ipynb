### Extracting Clips from this Youtube Video: https://www.youtube.com/watch?v=-uPc0yemm4k

```python
import sys
!{sys.executable} -m pip install yt-dlp moviepy --quiet

import yt_dlp
import os
from moviepy.video.io.ffmpeg_tools import ffmpeg_extract_subclip

# -----------------------------
# CONFIG
# -----------------------------
url = "https://www.youtube.com/watch?v=-uPc0yemm4k"

output_dir = "/Users/rianachatterjee/Downloads/in-lab/YOLO_VLM_switch_off/videos"
os.makedirs(output_dir, exist_ok=True)

full_video_path = os.path.join(output_dir, "temp_full_video.mp4")
clip_path = os.path.join(output_dir, "clip_28_00_to_29_00.mp4")

# -----------------------------
# Step 1: Download *lower-res* version (~720p max)
# -----------------------------
ydl_opts = {
    "format": "best[height<=720]",  # prevents huge 2.3GB download
    "outtmpl": full_video_path,
    "quiet": False
}

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    ydl.download([url])

print("Downloaded:", full_video_path)

# -----------------------------
# Step 2: Extract 28:00 → 29:00
# -----------------------------
start_time = 28 * 60
end_time = 29 * 60

ffmpeg_extract_subclip(full_video_path, start_time, end_time, targetname=clip_path)

print("Clip saved →", clip_path)
```

```python
# CELL 2: Video Clipping using ffmpeg-python
import ffmpeg

# Define file paths and times
input_file = "/Users/rianachatterjee/Downloads/in-lab/YOLO_VLM_switch_off/videos/temp_full_video.mp4"
output_file = "/Users/rianachatterjee/Downloads/in-lab/YOLO_VLM_switch_off/videos/clip_44_to_45_ffmpeg.mp4"

# Define the start and end times in seconds (28 minutes to 29 minutes)
start_time_seconds = 44 * 60  # 1680 seconds
duration_seconds = 1 * 60     # 60 seconds (1 minute clip)

print("Starting clip extraction using ffmpeg-python...")

try:
    # Use the input() function to specify the file and start time
    stream = ffmpeg.input(input_file, ss=start_time_seconds)
    
    # Use the output() function to specify the duration and output file
    stream = ffmpeg.output(
        stream, 
        output_file, 
        t=duration_seconds,          # Set the duration of the output clip
        vcodec='copy',               # Copy the video stream (fast, no re-encoding)
        acodec='copy'                # Copy the audio stream (fast, no re-encoding)
    )
    
    # Run the FFmpeg process
    ffmpeg.run(stream, overwrite_output=True)
    
    print(f"✅ Clip successfully extracted and saved to: {output_file}")
    
except ffmpeg.Error as e:
    print(f"❌ An FFmpeg error occurred. Details:")
    # Print the full command and stderr output for debugging
    print(e.cmd)
    print(e.stderr.decode('utf8'))
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
